```python
import pandas as pd
import numpy as np
```

```python
# Task 1 - Creating dataframes and saving as csv

    # Project DataFrame

project_df = pd.DataFrame({
    "ID": ["A001","A002","A003","A004","A005","A002","A005","A003","A001","A003","A001","A004","A003","A005"],
    "Project": ["Project 1","Project 2","Project 3","Project 4","Project 5","Project 6","Project 7","Project 8",
                "Project 9","Project 10","Project 11","Project 12","Project 13","Project 14"],
    "Cost": [1002000,2000000,4500000,5500000,np.nan,680000,400000,350000,np.nan,300000,2000000,1000000,300000,2000000],
    "Status": ["Finished","Ongoing","Finished","Ongoing","Finished","Failed","Finished","Failed",
               "Ongoing","Finished","Failed","Ongoing","Finished","Finished"]
})
    # Saving as csv

project_df.to_csv("project_df.csv")

    #Employee DataFrame

employee_df = pd.DataFrame({
    "ID": ["A001","A002","A003","A004","A005"],
    "Name": ["John Alter","Alice Luxumberg","Tom Sabestine","Nina Adgra","Amy Johny"],
    "Gender": ["M","F","M","F","F"],
    "City": ["Paris","London","Berlin","Newyork","Madrid"],
    "Age": [25,27,29,31,30]
})

    # Saving as csv
    
employee_df.to_csv("employee_df.csv")

    # Seniority level DataFrame

seniority_df = pd.DataFrame({
    "ID": ["A001","A002","A003","A004","A005"],
    "Designation Level": [2,2,3,2,3]
})

    # Saving as csv

seniority_df.to_csv("seniority_df.csv")
```

```python
# loading saved csv files
project_ld = pd.read_csv("project_df.csv", index_col=0)
employee_ld = pd.read_csv("employee_df.csv", index_col=0)
seniority_ld = pd.read_csv("seniority_df.csv", index_col=0)
```

```python
# Task 2 - reassining costs to blanks using running averages

for i in range(len(project_ld)):
    if pd.isna(project_ld.loc[i, "Cost"]):
        # Look at all previous rows in Cost column
        prev_values = project_ld.loc[:i-1, "Cost"].dropna()
        if len(prev_values) > 0:
            project_ld.loc[i, "Cost"] = prev_values.mean()
updated_values = project_ld.loc[4:8].drop(project_ld.index[5:8])
updated_values
```

```python
# Projetc DataFrame after replacing blanks
project_ld
```

```python
# Task 3 - Spliting name column of employee dataframe

    # Spliting Name column

employee_ld[["First Name","Last Name"]]= employee_ld["Name"].str.split(' ',expand=True, n=1)

    # deleting Name column

employee_ld = employee_ld.drop(columns=["Name"])

employee_ld
```

```python
# Task 4 - Joining all 3 dataframes into one

Final = project_ld.merge(employee_ld, on ="ID").merge(seniority_ld, on = "ID")
Final
```

```python
# Task 5 - Creating Bonus column

Final["Bonus"]= np.where(Final["Status"]=="Finished",Final["Cost"]*0.05,0)
Final
```

```python
# Sorting the Final DataFrame by ID in ascending order

Final = Final.sort_values(by="ID", ascending=True).reset_index(drop=True)

Final
```

```python
# Task 6 - Demoting employees with failed projects

    # Creating updated designation Level - row-wise

Final["Updated Designation Level"]= np.where(Final["Status"]=="Failed",Final["Designation Level"]-1,Final["Designation Level"])

    # Grouping the row-wise updated designation level to generate one updated designation level per employee

new_emp_lvl= Final.groupby("ID")["Updated Designation Level"].min().reset_index()

    # Joing the employee-wise updated designation level to the Final DataFrame

Final1 = Final.drop(columns=["Updated Designation Level"])
Final = Final1.merge(new_emp_lvl, on="ID", how="left")

# Code to drop level 4 (if any) in updated designation level

Final = Final[Final["Updated Designation Level"] != 4]

Final
```

```python
# Task 7 - Adding prefix to first name column and droping gender column

    # Adding Prefix

Final["First Name"] = np.where(Final["Gender"]=="M","Mr. "+Final["First Name"],"Mrs. "+Final["First Name"])

    # Dropping Gender column

Final = Final.drop(columns=["Gender"])

Final
```

```python
# Task 8 - Pormote all employee above 29 years by 1 level

Final["Updated Designation Level"] = Final.apply(
    lambda row: row["Updated Designation Level"] + 1 if row["Age"] > 29 else row["Updated Designation Level"],
    axis=1)
Final
```

```python
# Task 8 - Version 2 - where we have deleted level 4 as instructed in task 6

Final_without_4 = Final[Final["Updated Designation Level"] != 4]

Final_without_4
```

```python
# Task 9 - Total cost of projects per employee

TotalProjCost = pd.pivot_table(Final,index=["ID","First Name"],values="Cost")

# Renaming Cost column

TotalProjCost = TotalProjCost.rename(columns={"Cost": "Total Cost"})

TotalProjCost
```
